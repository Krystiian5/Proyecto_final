# Limpieza y Transformación de datos

En este notebook se aplica la limpieza del dataset `artists` mediante la función `clean_artists()` definida en `src/soporte_limpieza.py`.

Las principales transformaciones realizadas son:

- **id:** conversión a minúsculas.
- **name:** conversión a minúsculas, eliminación de espacios sobrantes y caracteres especiales innecesarios, manteniendo los valores nulos.
- **genres:** conversión del texto a listas de Python, normalización del contenido (minúsculas, eliminación de espacios y duplicados) y creación de la variable auxiliar `n_genres`.
- **main_genre:** conversión a minúsculas y eliminación de espacios sobrantes para unificar categorías.
- **Texto:** limpieza de caracteres Unicode inválidos para garantizar la correcta exportación del dataset en formato CSV.

```python
# ============================
# 1. Importaciones
# ============================

#Tratamiento de datos
import pandas as pd
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

#Exploracion de archivos
import sys
sys.path.append('../')

#Funciones personalizadas
from src import soporte_limpieza as spl 
```

```python
# ============================
# 2. Carga de datos
# ============================

df_o = pd.read_csv("../data/raw/artists.csv")
df = df_o.copy()
```

```python
# ============================
# 3. Limpieza y transformación
# ============================

df = spl.clean_artists(df)
```

```python
# ============================
# 4. Validaciones
# ============================

# Comprobar dimensiones
print(df.shape)

# Comprobar tipos de datos
display(df.dtypes)

# Comprobar nulos
display(df.isna().sum())

# Comprobar duplicados
print("Duplicados por ID:", df.duplicated(subset="id").sum())

# Revisar las nuevas columnas
display(
    df[
        [
            "genres",
            "main_genre",
            "n_genres"
        ]
    ].head()
)
```

```python
# ============================
# 5. Guardar dataset limpio
# ============================

df.to_csv("../data/processed/artists_clean.csv", index=False)
```
