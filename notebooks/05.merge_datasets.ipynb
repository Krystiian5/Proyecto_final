# Integración de datasets

En este notebook se integran los datasets limpios de canciones (`songs_clean.csv`) y artistas (`artists_clean.csv`) mediante la función `merge_songs_artists()` definida en `src/soporte_merge.py`.

El objetivo es construir un único dataset analítico que permita estudiar conjuntamente las características de las canciones y de sus artistas.

Durante este proceso se realizan las siguientes tareas:

- Carga de los datasets limpios.
- Conversión de la columna `artist_ids` en listas de Python.
- Expansión de las colaboraciones mediante `explode()`, generando una fila por artista.
- Unión de ambos datasets mediante el identificador del artista.
- Renombrado de columnas para mejorar la interpretación.
- Creación de variables auxiliares para el análisis.
- Validación del resultado del merge.
- Gestión de valores nulos.
- Exportación del dataset final.

```python
# ============================
# 1. Importaciones
# ============================
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)

import sys
sys.path.append("../")

from src import soporte_merge as spm
```

```python
# ============================
# 2. Carga de datos
# ============================
songs = pd.read_csv("../data/processed/songs_clean.csv")
artists = pd.read_csv("../data/processed/artists_clean.csv")
```

```python
# ============================
# 3. Integracióon de Datasets
# ============================
music = spm.merge_songs_artists(
    songs,
    artists
)
```

## Validación del merge

Una vez integrados ambos datasets, se realiza una validación inicial para comprobar que el proceso de unión se ha realizado correctamente y que el conjunto de datos resultante mantiene la calidad esperada.

En esta fase se revisan:

- Las dimensiones del dataset resultante.
- Los tipos de datos de cada variable.
- La existencia de valores nulos tras la integración.
- La presencia de registros duplicados considerando la combinación **song_id + artist_id**, ya que una misma canción puede estar asociada a varios artistas tras la expansión de colaboraciones.
- El número de canciones cuyo identificador de artista no encuentra correspondencia en el dataset de artistas.

Estas comprobaciones permiten detectar posibles inconsistencias antes de gestionar los valores nulos y generar el dataset final para el análisis exploratorio.

```python
# ============================
# 4. Validación del Merge
# ============================

print("Dimensiones del dataset:")
print(music.shape)

print("\nTipos de datos:")
display(music.dtypes)

print("\nValores nulos:")
display(
    music.isna()
         .sum()
         .sort_values(ascending=False)
)

print("\nDuplicados (song_id + artist_id):")
print(
    music.duplicated(
        subset=["song_id", "artist_id"]
    ).sum()
)

print("\nCanciones sin artista asociado:")
print(
    music["artist_name"].isna().sum()
)

display(music.head())
```

```python
# ============================
# 5. Revisión de nuevas variables
# ============================
display(
    music[
        [
            "song_name",
            "artist_name",
            "song_popularity",
            "artist_popularity",
            "artist_followers",
            "artist_followers_millions",
            "is_collaboration",
            "artist_has_genres"
        ]
    ].head()
)
```

## Gestión de valores nulos

Tras realizar el merge se revisan los valores nulos presentes en el dataset para decidir si deben conservarse o eliminarse.

No todos los valores nulos representan errores. En algunos casos corresponden a información que no está disponible en el dataset original y conservarlos resulta más apropiado que imputar valores artificiales.

Se adopta el siguiente criterio:

- **song_name:** se eliminan los registros sin nombre de canción, ya que impiden identificar correctamente el elemento analizado.

- **artist_name:** se eliminan los registros cuyo identificador de artista no existe en el dataset de artistas. Estos casos representan inconsistencias entre ambos conjuntos de datos y no permiten realizar análisis sobre el artista.

- **album_name:** se conservan los valores nulos, ya que algunas canciones corresponden a sencillos o el álbum no está informado en la fuente original.

- **main_artist:** se conservan los valores nulos porque proceden de listas vacías en el dataset original y no afectan al análisis al disponer igualmente del nombre del artista tras la integración.

- **popularity_per_million_followers:** se conservan los valores nulos porque la métrica no puede calcularse cuando el artista no dispone de seguidores registrados.

```python
# ============================
# 6. Gestión de Valores Nulos
# ============================
music = music.dropna(
    subset=["song_name", "artist_name"]
)
```

```python
# ============================
# 7. Validación Final
# ============================
print("Dimensiones finales:")
print(music.shape)

print("\nValores nulos restantes:")
display(
    music.isna().sum().sort_values(ascending=False)
)

print("\nDuplicados por canción y artista:")
print(
    music.duplicated(
        subset=["song_id", "artist_id"]
    ).sum()
)

display(music.head())
```

```python
# ============================
# 8. Guuardar Dataset
# ============================
music.to_csv(
    "../data/processed/music_dataset.csv",
    index=False
)
```
