# Integración de datasets

En este notebook se integran los datasets limpios de canciones (`songs_clean.csv`) y artistas (`artists_clean.csv`) mediante la función `merge_songs_artists()` definida en `src/soporte_merge.py`.

El objetivo es construir un único dataset analítico que permita estudiar conjuntamente las características de las canciones y de sus artistas.

Durante este proceso se realizan las siguientes tareas:

- Carga de los datasets limpios.
- Conversión de la columna `artist_ids` en listas de Python.
- Expansión de las colaboraciones mediante `explode()`, generando una fila por artista.
- Unión de ambos datasets mediante el identificador del artista.
- Renombrado de columnas para mejorar la interpretación.
- Creación de variables auxiliares para el análisis.
- Validación del resultado del merge.
- Gestión de valores nulos.
- Exportación del dataset final.

```python
# ============================
# 1. Importaciones
# ============================
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)

import sys
sys.path.append("../")

from src import soporte_merge as spm
```

```python
# ============================
# 2. Carga de datos
# ============================
songs = pd.read_csv("../data/processed/songs_clean.csv")
artists = pd.read_csv("../data/processed/artists_clean.csv")
```

```python
# ============================
# 3. Integración de Datasets
# ============================
music = spm.merge_songs_artists(
    songs,
    artists
)
```

## Validación del merge

Una vez integrados ambos datasets, se realiza una validación inicial para comprobar que el proceso de unión se ha realizado correctamente y que el conjunto de datos resultante mantiene la calidad esperada.

En esta fase se revisan:

- Las dimensiones del dataset resultante.
- Los tipos de datos de cada variable.
- La existencia de valores nulos tras la integración.
- La presencia de registros duplicados considerando la combinación **song_id + artist_id**, ya que una misma canción puede estar asociada a varios artistas tras la expansión de colaboraciones.
- El número de canciones cuyo identificador de artista no encuentra correspondencia en el dataset de artistas.

Estas comprobaciones permiten detectar posibles inconsistencias antes de gestionar los valores nulos y generar el dataset final para el análisis exploratorio.

```python
# ============================
# 4. Validación del Dataset Final
# ============================

print("Dimensiones del dataset:")
print(music.shape)

print("\nTipos de datos:")
display(music.dtypes)

print("\nValores nulos:")
display(
    music.isna()
         .sum()
         .sort_values(ascending=False)
)

print("\nCanciones duplicadas:")
print(
    music["song_id"].duplicated().sum()
)

print("\nCanciones únicas:")
print(
    music["song_id"].nunique()
)

print("\nColaboraciones:")
display(
    music["is_collaboration"].value_counts()
)

print("\nCanciones sin información de artista:")
print(
    music["main_artist"].isna().sum()
)

display(music.head())
```

```python
# ============================
# 5. Revisión de nuevas variables
# ============================

display(
    music[
        [
            "song_name",
            "main_artist",
            "song_popularity",
            "avg_artist_popularity",
            "total_artist_followers",
            "artist_followers_millions",
            "is_collaboration",
            "artist_has_genres"
        ]
    ].head()
)
```

```python
# ============================
# 6. Validación de colaboraciones
# ============================

print("Número de canciones colaborativas:")
print(
    music["is_collaboration"].sum()
)

print("\nPorcentaje de canciones colaborativas:")
print(
    round(
        music["is_collaboration"].mean() * 100,
        2
    ),
    "%"
)
```

## Gestión de valores nulos

Tras la preparación del dataset final se revisan los valores nulos presentes para determinar si deben conservarse o eliminarse.

No todos los valores nulos representan errores. En algunos casos corresponden a información que no está disponible en el dataset original, por lo que se considera más apropiado conservarlos que imputar valores artificiales.

Se adopta el siguiente criterio:

- song_name: se eliminan los registros sin nombre de canción, ya que impiden identificar correctamente la canción analizada. En total, se eliminan 40 registros.
- album_name: se conservan los valores nulos, ya que algunas canciones pueden no disponer de información sobre el álbum en la fuente original.
- main_artist: se conservan los valores nulos cuando no existe información disponible sobre el artista principal en el dataset original.
- popularity_per_million_followers: se conservan los valores nulos cuando no es posible calcular esta métrica a partir de la información disponible sobre los seguidores de los artistas.

El resto de variables no presentan valores nulos tras la preparación del dataset.

```python
# ============================
# 7. Gestión de Valores Nulos
# ============================

music = music.dropna(
    subset=["song_name"]
)
```

```python
# ============================
# 8. Validación Final
# ============================

print("Dimensiones finales:")
print(music.shape)

print("\nValores nulos restantes:")
display(
    music.isna()
         .sum()
         .sort_values(ascending=False)
)

print("\nCanciones duplicadas:")
print(
    music["song_id"].duplicated().sum()
)

print("\nCanciones únicas:")
print(
    music["song_id"].nunique()
)

print("\nDistribución de colaboraciones:")
display(
    music["is_collaboration"].value_counts()
)

print("\nPorcentaje de canciones colaborativas:")
print(
    round(
        music["is_collaboration"].mean() * 100,
        2
    ),
    "%"
)

display(music.head())
```

```python
# ============================
# 8. Guardar Dataset
# ============================
music.to_csv(
    "../data/processed/music_dataset.csv",
    index=False
)
```
